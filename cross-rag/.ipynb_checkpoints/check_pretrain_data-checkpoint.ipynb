# Pretrain 데이터베이스 Parquet 파일 확인

이 노트북은 pretrain 데이터베이스의 parquet 파일을 열어서 구조를 확인합니다.
특히 `distances`와 `indices`가 이미 계산되어 포함되어 있는지 확인합니다.

```python
import pandas as pd
import numpy as np
import os
from pathlib import Path

# Pretrain 데이터 디렉토리 경로
pretrain_data_dir = "../datasets/pretrain/pretrain_pairs_ctx512"

print(f"Pretrain 데이터 디렉토리: {pretrain_data_dir}")
print(f"디렉토리 존재 여부: {os.path.exists(pretrain_data_dir)}")

if os.path.exists(pretrain_data_dir):
    parquet_files = list(Path(pretrain_data_dir).glob("*.parquet"))
    print(f"\n발견된 parquet 파일 개수: {len(parquet_files)}")
    if len(parquet_files) > 0:
        print(f"첫 번째 파일: {parquet_files[0].name}")
        print(f"파일 크기: {parquet_files[0].stat().st_size / (1024*1024):.2f} MB")
else:
    print("디렉토리가 존재하지 않습니다!")
```

```python
# 첫 번째 parquet 파일 열기
if len(parquet_files) > 0:
    sample_file = parquet_files[0]
    print(f"파일 열기: {sample_file}")
    
    # Parquet 파일 읽기
    df = pd.read_parquet(sample_file)
    
    print(f"\n데이터프레임 shape: {df.shape}")
    print(f"\n컬럼 목록:")
    print(df.columns.tolist())
    
    print(f"\n데이터 타입:")
    print(df.dtypes)
    
    print(f"\n첫 5개 행:")
    df.head()
else:
    print("parquet 파일이 없습니다!")
```

```python
# distances와 indices 컬럼 확인
if len(parquet_files) > 0:
    print("=" * 60)
    print("distances와 indices 컬럼 확인")
    print("=" * 60)
    
    if 'distances' in df.columns:
        print("\n✓ 'distances' 컬럼이 존재합니다!")
        print(f"  - 데이터 타입: {df['distances'].dtype}")
        print(f"  - 샘플 데이터 (첫 번째 행):")
        sample_distances = df['distances'].iloc[0]
        if isinstance(sample_distances, (list, np.ndarray)):
            print(f"    타입: {type(sample_distances)}")
            print(f"    길이: {len(sample_distances)}")
            print(f"    첫 10개 값: {sample_distances[:10]}")
        else:
            print(f"    값: {sample_distances}")
    else:
        print("\n✗ 'distances' 컬럼이 존재하지 않습니다!")
    
    if 'indices' in df.columns:
        print("\n✓ 'indices' 컬럼이 존재합니다!")
        print(f"  - 데이터 타입: {df['indices'].dtype}")
        print(f"  - 샘플 데이터 (첫 번째 행):")
        sample_indices = df['indices'].iloc[0]
        if isinstance(sample_indices, (list, np.ndarray)):
            print(f"    타입: {type(sample_indices)}")
            print(f"    길이: {len(sample_indices)}")
            print(f"    첫 10개 값: {sample_indices[:10]}")
        else:
            print(f"    값: {sample_indices}")
    else:
        print("\n✗ 'indices' 컬럼이 존재하지 않습니다!")
    
    # target 컬럼도 확인
    if 'target' in df.columns:
        print("\n✓ 'target' 컬럼이 존재합니다!")
        print(f"  - 데이터 타입: {df['target'].dtype}")
        sample_target = df['target'].iloc[0]
        if isinstance(sample_target, (list, np.ndarray)):
            print(f"    타입: {type(sample_target)}")
            print(f"    길이: {len(sample_target)}")
            print(f"    첫 10개 값: {sample_target[:10]}")
        else:
            print(f"    값: {sample_target}")
```

```python
# 여러 파일 확인 (선택적)
print("=" * 60)
print("여러 파일에서 distances와 indices 일관성 확인")
print("=" * 60)

num_files_to_check = min(3, len(parquet_files))
print(f"\n{num_files_to_check}개 파일 확인 중...\n")

for i, file_path in enumerate(parquet_files[:num_files_to_check]):
    df_check = pd.read_parquet(file_path)
    print(f"파일 {i+1}: {file_path.name}")
    print(f"  - Shape: {df_check.shape}")
    print(f"  - 'distances' 존재: {'distances' in df_check.columns}")
    print(f"  - 'indices' 존재: {'indices' in df_check.columns}")
    print(f"  - 'target' 존재: {'target' in df_check.columns}")
    
    if 'distances' in df_check.columns and 'indices' in df_check.columns:
        # 첫 번째 행의 distances와 indices 길이 확인
        first_distances = df_check['distances'].iloc[0]
        first_indices = df_check['indices'].iloc[0]
        dist_len = len(first_distances) if isinstance(first_distances, (list, np.ndarray)) else 1
        idx_len = len(first_indices) if isinstance(first_indices, (list, np.ndarray)) else 1
        print(f"  - distances 길이 (첫 행): {dist_len}")
        print(f"  - indices 길이 (첫 행): {idx_len}")
    print()
```

```python
# 추가 분석: distances와 indices의 값 분포 확인
if len(parquet_files) > 0 and 'distances' in df.columns and 'indices' in df.columns:
    print("=" * 60)
    print("distances와 indices 값 분포 분석")
    print("=" * 60)
    
    # 첫 번째 행의 distances와 indices를 배열로 변환
    first_distances = np.array(df['distances'].iloc[0])
    first_indices = np.array(df['indices'].iloc[0])
    
    print(f"\n첫 번째 행 분석:")
    print(f"  - distances shape: {first_distances.shape}")
    print(f"  - distances min: {first_distances.min():.6f}, max: {first_distances.max():.6f}, mean: {first_distances.mean():.6f}")
    print(f"  - indices shape: {first_indices.shape}")
    print(f"  - indices min: {first_indices.min()}, max: {first_indices.max()}, mean: {first_indices.mean():.2f}")
    
    # 여러 행 확인
    print(f"\n여러 행 확인 (최대 5개):")
    for idx in range(min(5, len(df))):
        dists = np.array(df['distances'].iloc[idx])
        idxs = np.array(df['indices'].iloc[idx])
        print(f"  행 {idx}: distances 길이={len(dists)}, indices 길이={len(idxs)}")
        print(f"    distances 처음 5개: {dists[:5]}")
        print(f"    indices 처음 5개: {idxs[:5]}")
```

## 결론

위 결과를 통해 확인할 수 있는 것:
1. **distances와 indices가 이미 계산되어 포함되어 있는지**
2. **각 파일에서 이 값들이 일관되게 존재하는지**
3. **이 값들이 어떤 형태로 저장되어 있는지** (리스트, 배열 등)

만약 distances와 indices가 이미 포함되어 있다면, 모든 조합(similarity_space)에 대해 동일한 retrieval 결과를 사용하게 되어, 각 조합별로 다른 pretrain이 수행되지 않을 수 있습니다.

**문제점**: 
- 모든 조합(Z, X-cosine, X-euclidean, X-dtw 등)이 동일한 pretrain 데이터를 사용
- 각 조합별로 다른 retrieval 방식이 반영되지 않음
- 결과적으로 모든 조합의 pretrain checkpoint가 동일해질 수 있음
