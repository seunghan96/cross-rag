```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re

os.chdir('../ts-rag')
PATH = f'results_shift/forecast_evaluation'
settings = os.listdir(PATH)
settings = sorted(settings)
os.chdir(PATH)
```

```python
def parse_result_file(file_path):
    results_dict = {}  # Use a dictionary to keep only the last result per dataset
    with open(file_path, 'r') as f:
        lines = f.readlines()
    i = 0
    for idx in range(len(lines)):
        line = lines[idx].strip()
        if not line:
            continue
        if '_zeroshot_' in line:
            dataset = line.split('_512')[0]
            metric_line = lines[idx+1].strip()
            mse_match = re.search(r'mse:([\d.]+)', metric_line)
            mae_match = re.search(r'mae:([\d.]+)', metric_line)
            results_dict[dataset] = {
                'dataset': dataset,
                'mse': float(mse_match.group(1)),
                'mae': float(mae_match.group(1))
            }
    return pd.DataFrame(list(results_dict.values()))
```

```python
x = 'zeroshot_transfer_Chronos_lb512_X-cosine-norm_k15_TabPFN_dualhead_R_as_key_l0.7.txt'
```

```python
df = parse_result_file(x)

df['data1'] = df['dataset'].apply(lambda x:x.split('_')[0])
df['data2'] = df['dataset'].apply(lambda x:x.split('_')[3].replace('kb',''))
df.drop('dataset',axis=1,inplace=True)
```
