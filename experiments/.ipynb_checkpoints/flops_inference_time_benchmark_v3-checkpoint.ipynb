```python
def load_retrieved_data(dataset_name, root_dir, context_length, prediction_length, metadata, retrieve_suffix, use_z_space=False, embedding_model=None, embedding_tuning=None):
    """Load retrieved data from CSV file
    
    Args:
        use_z_space: If True, use retrieve_Z, else use retrieve_X_norm
        embedding_model: Required if use_z_space=True
        embedding_tuning: Optional, for Z-space retrieval
    """
    retrieval_database_names = '_'.join(metadata['database_name'])
    suffix = retrieve_suffix if retrieve_suffix else ''
    suffix = suffix if suffix == '' else f'_{suffix}'
    
    # Different file naming for Z-space vs X-space
    if use_z_space:
        embedding_tuning_str = embedding_tuning if embedding_tuning else 'None'
        retrieved_data_path = os.path.join(
            root_dir, 
            f'{dataset_name}_retrieve_{retrieval_database_names}_{metadata["lookback_length"]}_only_self_train_{embedding_tuning_str}{suffix}.csv'
        )
    else:
        retrieved_data_path = os.path.join(
            root_dir, 
            f'{dataset_name}_retrieve_{retrieval_database_names}_{metadata["lookback_length"]}_only_self_train_None{suffix}.csv'
        )
    
    #if not os.path.exists(retrieved_data_path):
    print(f'Retrieved data not found at {retrieved_data_path}')
    print('Performing retrieval...')
    # Perform retrieval
    if use_z_space:
        if embedding_model is None:
            raise ValueError("embedding_model is required for Z-space retrieval")
        retrieved_df = do_retrieve_Z(
            original_data_name=dataset_name,
            retrieval_database_dir=RETRIEVAL_DATABASE_DIR,
            root_dir=ROOT_DIR,
            metadata=metadata,
            mode='only_self_train',
            top_k=NUM_RETRIEVED,
            context_length=context_length,
            prediction_length=prediction_length,
            seed=SEED,
            dimension=768,  # Standard Chronos embedding dimension
            embedding_model=embedding_model,
            save=True,
            embedding_tuning=embedding_tuning,
            retrieve_suffix=retrieve_suffix
        )
    else:
        retrieved_df = do_retrieve_X_norm(
            original_data_name=dataset_name,
            retrieval_database_dir=RETRIEVAL_DATABASE_DIR,
            root_dir=ROOT_DIR,
            metadata=metadata,
            mode='only_self_train',
            top_k=NUM_RETRIEVED,
            context_length=context_length,
            prediction_length=prediction_length,
            seed=SEED,
            metric=METRIC,
            save=True,
            retrieve_suffix=retrieve_suffix
        )
#     else:
#         print(f'Loading retrieved data from {retrieved_data_path}')
#         retrieved_df = pd.read_csv(retrieved_data_path)
    
    return retrieved_df
```

```python
# Load embedding model for Z-space retrieval (ChronosBolt)
from chronos import ChronosPipeline

EMBEDDING_MODEL = None
EMBEDDING_TUNING = None  # Set to tuning path if using tuned embeddings

def get_embedding_model():
    """Get or create embedding model for Z-space retrieval"""
    global EMBEDDING_MODEL
    if EMBEDDING_MODEL is None:
        if EMBEDDING_TUNING is None:
            model_path = "amazon/chronos-t5-base"
        else:
            model_path = f"../tuning_results/{metadata['database_name'][0]}_{CONTEXT_LENGTH}_chronos_{EMBEDDING_TUNING}"
            if not os.path.exists(model_path):
                print(f'Warning: Tuned embedding model not found at {model_path}, using base model')
                model_path = "amazon/chronos-t5-base"
        
        print(f'Loading embedding model from {model_path}')
        EMBEDDING_MODEL = ChronosPipeline.from_pretrained(
            model_path,
            device_map='cuda' if torch.cuda.is_available() else 'cpu',
            torch_dtype=torch.bfloat16,
        )
    return EMBEDDING_MODEL
```

```python
def create_dummy_inputs(batch_size, context_length, num_retrieved, retrieved_seq_length):
    """Create dummy inputs for forward pass"""
    context = torch.randn(batch_size, context_length, device=device, dtype=torch.float32)
    retrieved_seq = torch.randn(batch_size, num_retrieved, retrieved_seq_length, device=device, dtype=torch.float32)
    distances = torch.randn(batch_size, num_retrieved, device=device, dtype=torch.float32)
    return context, retrieved_seq, distances

def measure_inference_time(model, context, retrieved_seq, distances, num_warmup=10, num_runs=100):
    """Measure inference time"""
    model.eval()
    
    # Warmup
    with torch.no_grad():
        for _ in range(num_warmup):
            _ = model(
                context=context,
                retrieved_seq=retrieved_seq,
                distances=distances
            )
    
    # Synchronize GPU
    if device.type == 'cuda':
        torch.cuda.synchronize()
    
    # Measure time
    times = []
    with torch.no_grad():
        for _ in range(num_runs):
            if device.type == 'cuda':
                torch.cuda.synchronize()
            start = time.time()
            
            _ = model(
                context=context,
                retrieved_seq=retrieved_seq,
                distances=distances
            )
            
            if device.type == 'cuda':
                torch.cuda.synchronize()
            end = time.time()
            times.append((end - start) * 1000)  # Convert to ms
    
    return np.mean(times), np.std(times), np.min(times), np.max(times)
```

```python
# ============================================================================
# Retrieval Efficiency Comparison: Z-space vs X-space (both using FAISS)
# ============================================================================
print("="*80)
print("RETRIEVAL EFFICIENCY COMPARISON")
print("="*80)
print("Comparing:")
print("  Z-space: Embedding computation + FAISS search")
print("  X-space: No embedding + FAISS search (on normalized raw time series)")
print("="*80)

retrieval_timings = {}

# Z-space retrieval (with embedding)
print("\n[1/2] Running Z-space retrieval (Embedding + FAISS)...")
if USE_ACTUAL_RETRIEVAL:
    try:
        embedding_model = get_embedding_model()
        retrieved_df_z_faiss, timing_z = do_retrieve_Z_faiss(
            original_data_name=DATASET_NAME,
            retrieval_database_dir=RETRIEVAL_DATABASE_DIR,
            root_dir=ROOT_DIR,
            metadata=metadata,
            mode='only_self_train',
            top_k=NUM_RETRIEVED,
            context_length=CONTEXT_LENGTH,
            prediction_length=PREDICTION_LENGTH,
            seed=SEED,
            dimension=768,
            embedding_model=embedding_model,
            save=True,
            embedding_tuning=EMBEDDING_TUNING,
            retrieve_suffix="Z_faiss"
        )
        retrieval_timings['Z-space'] = timing_z
        print(f"✓ Z-space retrieval completed")
    except Exception as e:
        print(f"✗ Z-space retrieval failed: {e}")
        import traceback
        traceback.print_exc()
        retrieval_timings['Z-space'] = None
else:
    print("Skipped (USE_ACTUAL_RETRIEVAL=False)")

# X-space retrieval (no embedding, FAISS only)
print("\n[2/2] Running X-space retrieval (No embedding + FAISS)...")
if USE_ACTUAL_RETRIEVAL:
    try:
        retrieved_df_x_faiss, timing_x = do_retrieve_X_faiss(
            original_data_name=DATASET_NAME,
            retrieval_database_dir=RETRIEVAL_DATABASE_DIR,
            root_dir=ROOT_DIR,
            metadata=metadata,
            mode='only_self_train',
            top_k=NUM_RETRIEVED,
            context_length=CONTEXT_LENGTH,
            prediction_length=PREDICTION_LENGTH,
            seed=SEED,
            metric=METRIC,
            save=True,
            retrieve_suffix="X_faiss"
        )
        retrieval_timings['X-space'] = timing_x
        print(f"✓ X-space retrieval completed")
    except Exception as e:
        print(f"✗ X-space retrieval failed: {e}")
        import traceback
        traceback.print_exc()
        retrieval_timings['X-space'] = None
else:
    print("Skipped (USE_ACTUAL_RETRIEVAL=False)")

# Display comparison
print("\n" + "="*80)
print("RETRIEVAL TIMING COMPARISON")
print("="*80)

if retrieval_timings.get('Z-space') and retrieval_timings.get('X-space'):
    timing_z = retrieval_timings['Z-space']
    timing_x = retrieval_timings['X-space']
    
    print(f"\nZ-space (Embedding + FAISS):")
    print(f"  DB Embedding time: {timing_z['embedding_time']:.2f} seconds")
    print(f"  FAISS Search time: {timing_z['search_time']:.2f} seconds")
    print(f"  Total time:        {timing_z['total_time']:.2f} seconds")
    
    print(f"\nX-space (No embedding + FAISS):")
    print(f"  DB Embedding time: {timing_x['embedding_time']:.2f} seconds (no embedding)")
    print(f"  FAISS Search time: {timing_x['search_time']:.2f} seconds")
    print(f"  Total time:        {timing_x['total_time']:.2f} seconds")
    
    print(f"\nSpeedup Analysis:")
    speedup_total = timing_z['total_time'] / timing_x['total_time'] if timing_x['total_time'] > 0 else 0
    speedup_search = timing_z['search_time'] / timing_x['search_time'] if timing_x['search_time'] > 0 else 0
    embedding_overhead = timing_z['embedding_time'] / timing_z['total_time'] * 100 if timing_z['total_time'] > 0 else 0
    
    print(f"  X-space is {speedup_total:.2f}x faster (total time)")
    print(f"  X-space is {speedup_search:.2f}x faster (search time only)")
    print(f"  Embedding overhead: {embedding_overhead:.1f}% of Z-space total time")
    
    # Store for later use
    retrieval_efficiency_results = {
        'Z-space': timing_z,
        'X-space': timing_x,
        'speedup_total': speedup_total,
        'speedup_search': speedup_search,
        'embedding_overhead_percent': embedding_overhead
    }
else:
    print("Timing comparison not available (one or both retrievals failed)")
    retrieval_efficiency_results = None

print("="*80)
```

```python
# Initialize models and measure
results = {}

for idx, (model_class, pipeline_class, model_name) in enumerate(zip(MODELS, PIPELINES, MODEL_NAMES)):
    print(f"\n{'='*60}")
    print(f"Processing model: {model_name}")
    print(f"{'='*60}")
    
    try:
        # Create model
        model = model_class(config=t5_config, augment=augment)
        model = model.to(device)
        model.eval()
        
        # Count parameters
        total_params = sum(p.numel() for p in model.parameters())
        trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
        
        print(f"Total parameters: {total_params:,}")
        print(f"Trainable parameters: {trainable_params:,}")
        
        # Create inputs - use dummy data for now (retrieved data integration can be added later if needed)
        # First 6 models use X-norm retrieval, ChronosBolt (idx=6) uses Z-space retrieval
        # For benchmarking purposes, dummy data is sufficient
        context, retrieved_seq, distances = create_dummy_inputs(
            BATCH_SIZE, CONTEXT_LENGTH, NUM_RETRIEVED, RETRIEVED_SEQ_LENGTH
        )
        
        # Measure FLOPS
        flops = None
        if HAS_THOP:
            try:
                # thop.profile expects a nn.Module object, not a function
                # It needs to register forward hooks, which requires .training attribute
                # Create a wrapper class that thop can hook into
                class ModelWrapper(nn.Module):
                    def __init__(self, base_model):
                        super().__init__()
                        self.model = base_model
                    
                    def forward(self, context, retrieved_seq, distances):
                        return self.model(
                            context=context,
                            retrieved_seq=retrieved_seq,
                            distances=distances
                        )
                
                wrapped_model = ModelWrapper(model)
                wrapped_model.eval()  # Ensure eval mode for FLOPS measurement
                
                flops, params = profile(
                    wrapped_model,
                    inputs=(context, retrieved_seq, distances),
                    verbose=False
                )
                flops_str = clever_format([flops], "%.3f")
                print(f"FLOPS: {flops_str}")
            except Exception as e:
                print(f"FLOPS measurement failed: {e}")
                import traceback
                traceback.print_exc()
        else:
            print("FLOPS measurement skipped (thop not available)")
        
        # Measure inference time
        mean_time, std_time, min_time, max_time = measure_inference_time(
            model, context, retrieved_seq, distances
        )
        
        print(f"Inference time: {mean_time:.3f} ± {std_time:.3f} ms (min: {min_time:.3f}, max: {max_time:.3f})")
        
        # Store results
        results[model_name] = {
            'total_params': total_params,
            'trainable_params': trainable_params,
            'flops': flops,
            'mean_time_ms': mean_time,
            'std_time_ms': std_time,
            'min_time_ms': min_time,
            'max_time_ms': max_time
        }
        
        # Clean up
        del model
        if device.type == 'cuda':
            torch.cuda.empty_cache()
            
    except Exception as e:
        print(f"Error processing {model_name}: {e}")
        import traceback
        traceback.print_exc()
        results[model_name] = {'error': str(e)}
```

```python
# Display results summary
print("\n" + "="*80)
print("SUMMARY")
print("="*80)

import pandas as pd

summary_data = []
for model_name, result in results.items():
    if 'error' not in result:
        summary_data.append({
            'Model': model_name,
            'Parameters': f"{result['total_params']:,}",
            'FLOPS': f"{result['flops']:,}" if result['flops'] else 'N/A',
            'Mean Time (ms)': f"{result['mean_time_ms']:.3f}",
            'Std Time (ms)': f"{result['std_time_ms']:.3f}",
            'Min Time (ms)': f"{result['min_time_ms']:.3f}",
            'Max Time (ms)': f"{result['max_time_ms']:.3f}",
        })
    else:
        summary_data.append({
            'Model': model_name,
            'Parameters': 'ERROR',
            'FLOPS': 'ERROR',
            'Mean Time (ms)': 'ERROR',
            'Std Time (ms)': 'ERROR',
            'Min Time (ms)': 'ERROR',
            'Max Time (ms)': 'ERROR',
        })

df = pd.DataFrame(summary_data)
print(df.to_string(index=False))
```

```python
# Visualization
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Parameters comparison
if all('error' not in results.get(m, {}) for m in MODEL_NAMES):
    params = [results[m]['total_params'] for m in MODEL_NAMES if 'error' not in results.get(m, {})]
    model_names_valid = [m for m in MODEL_NAMES if 'error' not in results.get(m, {})]
    axes[0].bar(model_names_valid, params)
    axes[0].set_ylabel('Number of Parameters')
    axes[0].set_title('Model Parameters')
    axes[0].tick_params(axis='x', rotation=45)
    
    # FLOPS comparison
    if HAS_THOP and all(results.get(m, {}).get('flops') for m in model_names_valid):
        flops = [results[m]['flops'] for m in model_names_valid]
        axes[1].bar(model_names_valid, flops)
        axes[1].set_ylabel('FLOPS')
        axes[1].set_title('FLOPS')
        axes[1].tick_params(axis='x', rotation=45)
    
    # Inference time comparison
    mean_times = [results[m]['mean_time_ms'] for m in model_names_valid]
    std_times = [results[m]['std_time_ms'] for m in model_names_valid]
    axes[2].bar(model_names_valid, mean_times, yerr=std_times, capsize=5)
    axes[2].set_ylabel('Inference Time (ms)')
    axes[2].set_title('Inference Time')
    axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('flops_inference_benchmark.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nPlot saved as 'flops_inference_benchmark.png'")
```

```python
# Save results to file
import json

# Convert numpy types to native Python types for JSON serialization
def convert_to_serializable(obj):
    if isinstance(obj, np.integer):
        return int(obj)
    elif isinstance(obj, np.floating):
        return float(obj)
    elif isinstance(obj, np.ndarray):
        return obj.tolist()
    elif isinstance(obj, dict):
        return {k: convert_to_serializable(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [convert_to_serializable(item) for item in obj]
    return obj

serializable_results = convert_to_serializable(results)

with open('flops_inference_results.json', 'w') as f:
    json.dump(serializable_results, f, indent=2)

print("Results saved to 'flops_inference_results.json'")
```
